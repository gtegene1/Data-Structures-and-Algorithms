```python
"""Write a short Python function, is multiple(n, m), that takes two integer values and returns True if n is a multiple of m, that is, n = mi for some
integer i, and False otherwise."""

def is_multiple(n,m):
    if n/m == int(n/m):
        return True
    elif m/n == int(m/n):
        return True
    else:
        return False
    
is_multiple(20,10)
```

```python
"""Write a function that returns the sum of all even positive integers smaller than n."""

def sum_evens(n):
    x = 0
    n = n - 1 
    while n > 0:
     if n%2 == 0:
       x = x + n
       n = n - 1
     else:
        n = n - 1
    print(x)
sum_evens(8)
```

```python
""" Sum of Cubes of Numbers Below n
Return the sum of cubes of all numbers smaller than n."""

def sum_cubes(z):
    l = 0
    z = z - 1
    while z > 0 :
        l = l + z**3
        z = z - 1
    print(l)
sum_cubes(5)
```

```python
""" Sum of Squares Until Limit
Keep adding squares (1² + 2² + 3² + …) until the sum is ≥ n.
Return how many numbers you used. """

def count_squares(c):
    num = 0
    interm = 0
    for i in range(1,c):
        while i >interm:
            interm = interm + i**2
            num = num + 1
    print(num)      

count_squares(20)
```
