```python
"""
Product of a List
Given a list of numbers, return the product of all elements using recursion.
product([2, 3, 4]) → 24

"""
def product(x):
    if len(x) == 1:
        return x[0]
    else:
        return x[0] * product(x[1:])
x=[2,3,4]
product(x)


```

```python
"""
2. Count Occurrences

Given a list and a target value, return how many times the target appears.

count_occurrences([1, 2, 3, 2, 2], 2) → 3
"""

def count_occurrences(x,y):
    if len(x) == 0:
        return 0
    else:
        if x[0] == y:
            return 1 + count_occurrences(x[1:],y)
        else: return count_occurrences(x[1:],y)

count_occurrences([1, 2, 3, 2, 2], 2)
```

```python
"""
3. Sum of Even Numbers

Given a list of integers, return the sum of only the even numbers.

sum_evens([1, 2, 3, 4, 5, 6]) → 12
"""

def sum_evens(x):
    if len(x) == 0:
        return 0
    else:
        if (x[0] % 2) == 0:
            return x[0] + sum_evens(x[1::])
        else : return sum_evens(x[1::])


x = [1, 2, 3, 4, 5, 10]
sum_evens(x)
```

```python
"""
5. Find the Minimum

Given a non-empty list, return the smallest value without using min().

find_min([3, 1, 4, 1, 5, 9]) → 1

"""
def find_min(x):
    if len(x) == 1:
        return x[0]
    else:
        if x[0] > find_min(x[1:]):
            return find_min(x[1:])
        else:
            return x[0]

x = [3,-1,4,1,5,8,9]

find_min(x)
```

```python
"""

6. Power of Two Check

Given a positive integer n, return True if it's a power of 2, False otherwise. No loops, no math module.

is_power_of_two(8) → True, is_power_of_two(6) → False
"""  

def is_power_of_two(x):
    if x == 1:  ## 2 sqrt 1 is 2
        return True
    else:
        if (x % 2) == 0:
            return is_power_of_two(x/2)
        else:
            return False
        
is_power_of_two(1)
```

```python
"""
7. Remove Duplicates

Given a list of integers, return a new list with duplicates removed, preserving order.

remove_dupes([1, 2, 2, 3, 1, 4]) → [1, 2, 3, 4]
"""

def remove_dupes(x):
    if len(x) == 1:
        return x
    else:
        if x[0] in x[1:] :
           return remove_dupes(x[1:])
        else:
            return  [x[0]] + remove_dupes(x[1:])


remove_dupes([1, 2, 2, 3, 1, 4])
```

```python
## A list can contain integers or other lists (arbitrarily deep). Return the sum of all integers.

x = [1, [2, [3, 4]], 5]

def nested_sum(z):
    if z == [] :
        return 0
    elif isinstance(z[0],list) == False :
        return z[0] + nested_sum(z[1:])
    else:
        return nested_sum(z[0]) + nested_sum(z[1:])



nested_sum(x)
```

```python
"""

Binary Search

Given a sorted list and a target, return the index of the target (or -1 if not found). No loops.

binary_search([1, 3, 5, 7, 9], 7) → 3

"""

x = [1,3,5,7,9]

def binary_search(a,b, low, high):

    mid = (low + high)//2
    if low > high:
        return -1
    elif a[mid] == b :
        return mid
    elif a[mid] > b:
        return binary_search(a,b,low,  mid - 1)
    elif a[mid] <b:
        return  binary_search(a,b, mid+1, high)



binary_search(x,7,0, len(x) -1 )
```

```python
"""
1. Sum of Digits (recursive)

Given a non-negative integer, return the sum of its digits.

digit_sum(1234) → 10
💡 Hint: n % 10 gives the last digit. n // 10 chops it off.

"""

def digit_sum(x):
    if x // 10 == 0:
        return x
    else:
      return x % 10  + digit_sum(x//10)

digit_sum(3056)
```
